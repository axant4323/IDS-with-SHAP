```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining label distribution:")
print(y_train.value_counts())

print("\nTesting label distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["proto", "service", "state"]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ],
    remainder="passthrough"
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Original features:", X_train.shape[1])
print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
y_pred = rf_model.predict(X_test_processed)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
```

# 5. Advanced Pipeline & Performance Improvements

To address the limitations of the baseline models (namely, severe class imbalance in multiclass, high-cardinality categorical sparsity, and lack of domain-specific flow indicators), we introduce:
1. **Network Flow Feature Engineering**:
   - Volume and asymmetry ratios (`bytes_ratio`, `pkts_ratio`, `sbytes_per_pkt`, `dbytes_per_pkt`, `byte_diff`, `pkt_diff`).
   - Rate and load dynamics (`byte_rate`, `pkt_rate`, `sload_to_rate`, `dload_to_rate`, `load_ratio`).
   - TCP handshake, jitter, and inter-packet timing (`tcprtt_to_dur`, `syn_ack_diff`, `dinpkt_to_sinpkt`, `jit_diff`, `jit_ratio`, `jit_sum`).
   - Connection & window dynamics (`ct_ratio`, `swin_dwin_ratio`, `stcpb_dtcpb_diff`, `has_tcp_handshake`, `is_slow_handshake`).
   - Log1p scaling on heavy-tailed distributions (`dur`, `sbytes`, `dbytes`, `sload`, `dload`, `rate`, `sinpkt`, `dinpkt`, `sjit`, `djit`).
   - Categorical frequency encoding and rare-category pooling (`min_frequency=50`).
2. **Cost-Sensitive Power Class Weighting**:
   - Smoothed weighting $(N / (C \cdot N_c))^lpha$ with $lpha = 0.40$ to give gradient attention to minority attack categories (`Worms`, `Backdoor`, `Analysis`, `DoS`, `Shellcode`) without inflating false positive rates on `Normal`.
3. **Advanced Gradient Boosters & Soft-Voting Ensembles**:
   - Tuned **LightGBM** (histogram-based boosting with high leaf capacity).
   - Tuned **XGBoost** with regularized depth and tree subsampling.
   - **Soft-Voting Ensemble** combining calibrated class posteriors.
4. **Decision Threshold Optimization** for Binary Classification.

```python
# -------------------------------------------------------------
# 5.1 Domain Feature Engineering Function
# -------------------------------------------------------------
def build_network_features(df_in):
    """
    Constructs domain-specific flow indicators, timing ratios,
    asymmetry metrics, and log-scaled variables for network intrusion detection.
    """
    d = df_in.copy()
    
    # Flow volumes & totals
    d['tot_bytes'] = d['sbytes'] + d['dbytes']
    d['tot_pkts'] = d['spkts'] + d['dpkts']
    d['tot_loss'] = d['sloss'] + d['dloss']
    d['byte_diff'] = np.abs(d['sbytes'] - d['dbytes'])
    d['pkt_diff'] = np.abs(d['spkts'] - d['dpkts'])
    
    # Asymmetric flow ratios
    d['bytes_ratio'] = (d['sbytes'] + 1.0) / (d['dbytes'] + 1.0)
    d['pkts_ratio'] = (d['spkts'] + 1.0) / (d['dpkts'] + 1.0)
    d['sbytes_per_pkt'] = d['sbytes'] / (d['spkts'] + 1.0)
    d['dbytes_per_pkt'] = d['dbytes'] / (d['dpkts'] + 1.0)
    d['loss_ratio'] = d['tot_loss'] / (d['tot_pkts'] + 1.0)
    d['sloss_rate'] = d['sloss'] / (d['spkts'] + 1.0)
    d['dloss_rate'] = d['dloss'] / (d['dpkts'] + 1.0)
    
    # Rates and load dynamics
    d['byte_rate'] = d['tot_bytes'] / (d['dur'] + 1e-5)
    d['pkt_rate'] = d['tot_pkts'] / (d['dur'] + 1e-5)
    d['sload_to_rate'] = d['sload'] / (d['rate'] + 1e-5)
    d['dload_to_rate'] = d['dload'] / (d['rate'] + 1e-5)
    d['load_ratio'] = (d['sload'] + 1.0) / (d['dload'] + 1.0)
    
    # TCP Handshake, jitter, and inter-packet timing
    d['tcprtt_to_dur'] = d['tcprtt'] / (d['dur'] + 1e-5)
    d['synack_to_ackdat'] = (d['synack'] + 1e-5) / (d['ackdat'] + 1e-5)
    d['tcprtt_synack_ratio'] = d['synack'] / (d['tcprtt'] + 1e-5)
    d['syn_ack_diff'] = np.abs(d['synack'] - d['ackdat'])
    d['dinpkt_to_sinpkt'] = (d['dinpkt'] + 1e-4) / (d['sinpkt'] + 1e-4)
    d['jit_diff'] = np.abs(d['sjit'] - d['djit'])
    d['jit_ratio'] = (d['sjit'] + 1.0) / (d['djit'] + 1.0)
    d['jit_sum'] = d['sjit'] + d['djit']
    d['dinpkt_dload_ratio'] = d['dinpkt'] / (d['dload'] + 1e-4)
    d['mean_size_ratio'] = (d['smean'] + 1.0) / (d['dmean'] + 1.0)
    
    # Connection counts and windows
    d['ct_ratio'] = (d['ct_src_dport_ltm'] + 1.0) / (d['ct_dst_sport_ltm'] + 1.0)
    d['swin_dwin_ratio'] = (d['swin'] + 1.0) / (d['dwin'] + 1.0)
    d['stcpb_dtcpb_diff'] = np.abs(d['stcpb'] - d['dtcpb'])
    
    # Boolean indicator flags
    d['has_tcp_handshake'] = (d['tcprtt'] > 0).astype(int)
    d['is_slow_handshake'] = (d['tcprtt'] > 0.05).astype(int)
    d['zero_dbytes'] = (d['dbytes'] == 0).astype(int)
    d['zero_dur'] = (d['dur'] == 0).astype(int)
    
    # Log transforms for skewed flow variables
    skew_cols = ['sbytes', 'dbytes', 'tot_bytes', 'sload', 'dload', 'rate', 'dur', 'sinpkt', 'dinpkt', 'sjit', 'djit']
    for c in skew_cols:
        d[f'log_{c}'] = np.log1p(np.maximum(0, d[c]))
        
    return d

# Apply feature engineering to cleaned model dataset
df_enhanced = build_network_features(df_model)

# Frequency encoding for high-cardinality categoricals
cat_cols = ['proto', 'service', 'state']
for col in cat_cols:
    freq_map = df_enhanced[col].value_counts(normalize=True)
    df_enhanced[f'{col}_freq'] = df_enhanced[col].map(freq_map)

print(f"Features before engineering: {df_model.shape[1]}")
print(f"Features after engineering : {df_enhanced.shape[1]}")
```

### 5.2 Upgraded Binary Classification Pipeline
We train:
1. **Tuned XGBoost** with histogram tree method, depth 7, subsampling 0.8.
2. **Tuned LightGBM** with 127 leaves and min_child_samples 20.
3. **Soft-Voting Ensemble** combining both models.
4. **Decision Threshold Optimization** scanning thresholds from 0.35 to 0.60.

```python
import lightgbm as lgb
from sklearn.metrics import roc_auc_score

# Binary train/test split with exact same random state
X_bin_adv = df_enhanced.drop(columns=['label', 'attack_cat'])
y_bin_adv = df_enhanced['label']

X_tr_b, X_te_b, y_tr_b, y_te_b = train_test_split(
    X_bin_adv, y_bin_adv, test_size=0.20, random_state=42, stratify=y_bin_adv
)

# ColumnTransformer with min_frequency=50 to eliminate rare-category sparsity
ct_binary = ColumnTransformer(
    transformers=[('cat', OneHotEncoder(min_frequency=50, handle_unknown='ignore', sparse_output=False), cat_cols)],
    remainder='passthrough'
)
X_tr_b_proc = ct_binary.fit_transform(X_tr_b)
X_te_b_proc = ct_binary.transform(X_te_b)

# 1. Tuned XGBoost
xgb_bin_tuned = XGBClassifier(
    n_estimators=450,
    max_depth=7,
    learning_rate=0.04,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    tree_method='hist'
)
xgb_bin_tuned.fit(X_tr_b_proc, y_tr_b)
xgb_bin_probs = xgb_bin_tuned.predict_proba(X_te_b_proc)[:, 1]

# 2. Tuned LightGBM
lgb_bin_tuned = lgb.LGBMClassifier(
    n_estimators=500,
    learning_rate=0.04,
    num_leaves=127,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_samples=20,
    random_state=42,
    n_jobs=-1,
    verbose=-1
)
lgb_bin_tuned.fit(X_tr_b_proc, y_tr_b)
lgb_bin_probs = lgb_bin_tuned.predict_proba(X_te_b_proc)[:, 1]

# 3. Soft-Voting Ensemble
ensemble_bin_probs = 0.5 * xgb_bin_probs + 0.5 * lgb_bin_probs
y_pred_ensemble_b = (ensemble_bin_probs >= 0.50).astype(int)

print("Tuned Binary Models training completed.")
```

```python
# Threshold optimization for Binary Ensemble
threshold_rows = []
for t in np.arange(0.35, 0.65, 0.05):
    preds_t = (ensemble_bin_probs >= t).astype(int)
    threshold_rows.append({
        "Threshold": round(t, 2),
        "Accuracy": accuracy_score(y_te_b, preds_t),
        "Precision": precision_score(y_te_b, preds_t),
        "Recall": recall_score(y_te_b, preds_t),
        "F1-Score": f1_score(y_te_b, preds_t)
    })

thresh_df = pd.DataFrame(threshold_rows)
print("=== Decision Threshold Scan for Binary Ensemble ===")
display(thresh_df)

best_thresh_idx = thresh_df["F1-Score"].idxmax()
print(f"Optimal F1 Threshold: {thresh_df.loc[best_thresh_idx, 'Threshold']} (F1 = {thresh_df.loc[best_thresh_idx, 'F1-Score']:.4f})")
```

### 5.3 Upgraded Multiclass Classification Pipeline
To address the severe class imbalance where minority classes (`Worms`, `Backdoor`, `Analysis`, `DoS`) are otherwise drowned out by `Normal` (63.3%), we apply:
- **Cost-Sensitive Power Class Weighting**:
  $$w_c = \left(\frac{N}{C \cdot N_c}\right)^{0.40}$$
- **Tuned Multiclass XGBoost** & **Tuned Multiclass LightGBM**
- **Multiclass Soft-Voting Ensemble**

```python
# Multiclass preparation with enhanced features
df_multi_adv = df_enhanced.copy()
df_multi_adv["attack_cat"] = df_multi_adv["attack_cat"].astype("string").str.strip()
df_multi_adv = df_multi_adv.dropna(subset=["attack_cat"]).copy()

X_m_adv = df_multi_adv.drop(columns=["label", "attack_cat"])
y_m_adv_text = df_multi_adv["attack_cat"]

le_adv = LabelEncoder()
y_m_adv = le_adv.fit_transform(y_m_adv_text)
class_names_adv = le_adv.classes_

X_tr_m, X_te_m, y_tr_m, y_te_m = train_test_split(
    X_m_adv, y_m_adv, test_size=0.20, random_state=42, stratify=y_m_adv
)

# OneHotEncoder with min_frequency=50
ct_multi = ColumnTransformer(
    transformers=[('cat', OneHotEncoder(min_frequency=50, handle_unknown='ignore', sparse_output=False), cat_cols)],
    remainder='passthrough'
)
X_tr_m_proc = ct_multi.fit_transform(X_tr_m)
X_te_m_proc = ct_multi.transform(X_te_m)

# Compute power class weights (alpha=0.40)
classes_arr, counts_arr = np.unique(y_tr_m, return_counts=True)
n_total = len(y_tr_m)
n_cls = len(classes_arr)
weights_power = (n_total / (n_cls * counts_arr)) ** 0.40
weight_map = dict(zip(classes_arr, weights_power))
sw_train = np.array([weight_map[c] for c in y_tr_m])

print("Class weighting distribution:")
for c_idx, c_name in enumerate(class_names_adv):
    print(f"  {c_name:15s}: count={counts_arr[c_idx]:5d}, weight={weight_map[c_idx]:.3f}")
```

```python
# 1. Upgraded Multiclass XGBoost with sample weights
xgb_multi_adv = XGBClassifier(
    n_estimators=350,
    max_depth=7,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="multi:softprob",
    num_class=len(class_names_adv),
    eval_metric="mlogloss",
    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)
xgb_multi_adv.fit(X_tr_m_proc, y_tr_m, sample_weight=sw_train)
xgb_multi_probs = xgb_multi_adv.predict_proba(X_te_m_proc)

# 2. Upgraded Multiclass LightGBM with sample weights
lgb_multi_adv = lgb.LGBMClassifier(
    n_estimators=400,
    learning_rate=0.05,
    num_leaves=95,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_samples=15,
    random_state=42,
    n_jobs=-1,
    verbose=-1
)
lgb_multi_adv.fit(X_tr_m_proc, y_tr_m, sample_weight=sw_train)
lgb_multi_probs = lgb_multi_adv.predict_proba(X_te_m_proc)

# 3. Soft-Voting Ensemble
ensemble_multi_probs = 0.5 * xgb_multi_probs + 0.5 * lgb_multi_probs
y_pred_ensemble_multi = np.argmax(ensemble_multi_probs, axis=1)

print("Multiclass Ensemble training completed.")
```

```python
# Multiclass Ensemble Evaluation
print(f"Ensemble Accuracy   : {accuracy_score(y_te_m, y_pred_ensemble_multi):.4f}")
print(f"Ensemble Macro F1   : {f1_score(y_te_m, y_pred_ensemble_multi, average='macro', zero_division=0):.4f}")
print(f"Ensemble Weighted F1: {f1_score(y_te_m, y_pred_ensemble_multi, average='weighted', zero_division=0):.4f}")

print("\nMulticlass Ensemble Classification Report:\n")
print(classification_report(
    y_te_m,
    y_pred_ensemble_multi,
    labels=range(len(class_names_adv)),
    target_names=class_names_adv,
    zero_division=0
))
```
